```python
def mean(arr):
    n=len(arr)
    return np.sum(arr)/n
```

```python
mean(np.array([1,2,3,4,5,6]))
```

```python
def variance(array):
    m=mean(array)
    n=len(array)
    return np.sum((array-m)**2)/n
```

```python
variance(np.array([1,2,3,4,5,6]))
```

```python
def std(array):
    v=variance(array)
    return np.sqrt(v)
```

```python
std(np.array([1,2,3,4,5,6]))
```

```python
def minmax(array):
    a=np.array([1,2,3,4,5,6])
    xmin=np.min(a)
    xmax=np.max(a)
    result=((a-xmin)/(xmax-xmin))
    return result
```

```python
minmax(np.array([1,2,3,4,5,6]))
```

```python
import numpy as np

def r(x):
    x = np.array(x)
    y= np.maximum(0, x)
    return y
```

```python
r(np.array([1,2,3,4,5,6]))
```

```python
def sigmoid(x):
    y=1 / (1 + np.exp(-x))
    return y
```

```python
sigmoid(np.array([1,2,3,4,5,6]))
```
